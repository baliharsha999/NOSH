```python
import json, glob
import pandas as pd

frames = []
for path in glob.glob("data_recipes/*.json"):
    raw = json.load(open(path))
    f = pd.DataFrame.from_dict(raw, orient="index") 
    f.index.name = "id"
    f = f.reset_index()
    f["source_file"] = path
    frames.append(f)

df_raw = pd.concat(frames, ignore_index=True)
df_raw["id"] = df_raw["id"].astype(str)

print(df_raw.shape, df_raw.columns.tolist())
print("duplicate ids across files:", df_raw.id.duplicated().sum())
print(df_raw.isna().sum())
```
